```python
avg_G3 = df.groupby("studytime")["G3"].mean()
```

```python
plt.figure(figsize = (8,5))
avg_G3.plot(kind="bar")
plt.title("Average of G3 with studytime",color="b")
plt.xlabel("studytime",color="r")
plt.ylabel("Average of G3",color="r")
plt.grid()
plt.show()
```

```python
avg_sex= df.groupby("sex")["G3"].mean()
print(avg_sex)
```

```python
plt.figure(figsize =(6,4))
avg_sex.plot(kind="bar")
plt.grid()
plt.title("Average of sex")
plt.xlabel("sex")
plt.ylabel("G3")
plt.show()
```

```python
avg_age=df.groupby("age")["G3"].mean()
print(avg_age)
```

```python
plt.figure(figsize =(6,4))
avg_age.plot(kind="bar")

plt.title("Average of age by G3")
plt.xlabel("age")
plt.ylabel("G3")
plt.show()
```

```python
df['result']= df['G3'].apply(lambda x:"Pass" if x>=10 else "Fail")
```

```python
labels = ['Pass','Fail']
sizes = [265,130]
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    wedgeprops={"width":0.4}
)
plt.title("Count of result")
plt.show()
```

```python
avg_Medu= df.groupby('Medu')['G3'].mean()
print(avg_Medu)
```

```python
plt.figure(figsize=(7,5))
plt.bar(
    avg_Medu.index,
    avg_Medu.values,
    color="#4CAF50"
)
plt.title("Average G3 Marks by Mother's Education")
plt.xlabel("Mother's Education (Medu)")
plt.ylabel("Average G3 Marks")
plt.show()
```

# This Student Performance Analysis project explored the factors affecting students' academic performance using Python and Pandas. The dataset was cleaned, analyzed, and visualized using various charts.

Key findings:
• Most students successfully passed the final examination.
• Student performance varies based on factors such as study time, parental education, and family support.
• Higher study time generally leads to better final grades.
• Students with greater parental educational support tend to achieve higher average scores.
• Data visualization using bar charts, donut charts, and other graphs made the patterns easier to understand.

Overall, this project demonstrates how data analysis can help identify the factors influencing student performance and support better educational decision-making. It also strengthened practical skills in Python, Pandas, Matplotlib, and exploratory data analysis (EDA).
